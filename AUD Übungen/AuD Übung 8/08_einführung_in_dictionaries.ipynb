# Dictionaries/ Hash-Table und Sets in Python

### Dictionaries

#### Erstellen

```python
dic = dict() # Erstellen eines leeren Dictionaries/ Hash-Tabellen 
dic = {}    # Erstellen eines leeren Dictionaries/ Hash-Tabellen 
dic = {'a':12, 'b': 13} # Erstellen einer Tabelle mit zwei Schlüssel-Wert-Paaren
```

#### Lesen

```python
dic.get('c') # Sichererweg Werte zu Lesen (gibt None zurück wenn Schlüssel nicht enthalten)
dic['b'] # Zugriff auf Elemente
```

#### Schreien/ Hinzufügen

```python
dic['c']=15 # Ändern von Werten für vorhandene Schlüssel oder Hinzufügen von Schlüssel-Wert-Paar für neue Schlüssel
dic.update({'d':122,'e':14}) # Ändern von Werten für vorhandene Schlüssel oder Hinzufügen von Schlüssel-Wert-Paaren für neue Schlüssel
```

#### Iterieren

```python
dic={"a":6, "b":7, "c":8}
for key in dic: # Iterieren über die Elemente im Dictionary
    print(key, ',', dic[key]) # Ausgabe der Schlüssel und Werte
```

### Sets

#### Erstellen

```python
s = set() # Erstellen eines leeren Sets
s = {12, 13, 14} # Erstellen eines Sets mit drei Elementen
```

#### Lesen

```python
print(14 in s) # Testen ob eine Element im Set ist
```

#### Hinzufügen

```python
s.add(14) # Hinzufügen eines Elements, falls es noch nicht enthalten ist
```

#### Iterieren 

```python
for element in s: # Iterieren über die Elemente im Set
    print(element) # Ausgabe der Elemente
```
