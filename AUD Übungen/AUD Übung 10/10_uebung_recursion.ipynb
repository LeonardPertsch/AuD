b) Nutze Tiefensuche, um einen Algorithmus zu gestalten, der ermittelt, ob zwei beliebige Knoten
von $G$ Teil derselben starken Zusammenhangskomponente sind. Implementiere deinen Algorithmus unter Verwendung der Funktion `tiefensuche` und teste ihn für die Knoten $a$ und $c$.

```python
graph1 = {"a": {"b","d"},
         "b": {"a"},
         "c": {"a"},
         "d": {"c"},
         "e": {"f"},
         "f": {"e"}
        }
```

```python
def tiefensuche (graph, visited, v1, v2):
    visited.add(v1)
    if v1 == v2:
        return True
    for u in graph[v1].difference(visited):
        if tiefensuche (graph, visited, u, v2):
            return True
    return False
```

```python
def connected(graph, v1, v2):
    visited1 = set()
    
    visited2 = set()
    return tiefensuche(graph, visited2, v2, v1) and tiefensuche(graph, visited1, v1, v2)
    
```

```python
print(connected(graph1, "a", "c"))
print(connected(graph1, "a", "e"))
```

```python
def zusammenhangskomponente (graph,komponente, v):
    komponente.add(v)
    toVisit = []
    for e in graph[v]:
        toVisit.append(v)
    while len(toVisit) > 0:
        u = toVisit.pop(0)
        komponente.add(u)
        for e in graph[u].difference(komponente):
            toVisit.append(e)
        
```
