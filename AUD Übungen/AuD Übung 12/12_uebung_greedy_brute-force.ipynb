## Aufgabe 1: Wechselgeld-Algorithmus

a) Implementiere einen Algorithmus für einen Bezahlautomaten, der als Eingabe die Kosten des
Einkaufs und den vom Kunden gegebenen Geldbetrag nimmt. Der Algorithmus soll eine Stückelung des herauszugebenden Wechselgeldes in die einzelnen Euromünzen berechnen, wobei
es aus Gründen der Kundenfreundlichkeit folgendes zu beachten gilt:
- Die Anzahl der herausgegebenen Münzen soll so klein wie möglich sein.
- Die Laufzeit des Algorithmus soll so kurz wie möglich sein.

```python
from numpy.matlib import empty


def wechselgeld(kosten, gegeben):
    betrag = (gegeben - kosten) * 100
    out=[]
    muenzen = [200, 100, 50, 20 ,10,5,2,1]
    i = 0
    while betrag>0 and i<len(muenzen):
        current = muenzen[i]
        if current > betrag:
            i += 1
        else:
            betrag = betrag- current
            out.append(current/100)
    return out
```

## Aufgabe 2
Dein Nachbar nutzt einen alten WLAN-Router, der nur Zahlen als Passwörter unterstützt. Nachdem du eines seiner Netzwerkpakete abgefangen hast, findest du nach etwas Recherche auch die Implementierung der Paketverschlüsselung des Routermodells. Welche Art von Inhalten konsumiert dein Nachbar im Internet?

*Hinweis: Die Kopfzeile von HTTP-Anfragen beginnt mit GET und endet mit HTTP/1.1.*

*Die Verschlüsselung ist symmetrisch, d.h. Ver- und Entschlüsselung nutzen dasselbe Passwort.*

```python
def encrypt(text, password):
    encrypted = ""
    for i in range(len(text)):
        encrypted += chr(ord(text[i]) ^ ord(password[i % len(password)])) # ^ ist bitweises XOR
    return encrypted

def decrypt(text, password):
    
    return encrypt(text, password)  # XOR ist symmetrisch
```

Das abgefangene Paket wird hier in der Variablen packet gespeichert (es enthält nicht-druckbare Zeichen).

```python
packet = "".join(map(chr, [113, 113, 97, 22, 29, 89, 87, 64, 79, 83, 92, 80, 95, 88, 81, 83, 64, 28, 76, 93, 69, 22, 122, 102, 98, 100, 26, 7, 28, 3]))
print(packet)
```

```python
i = 0
cracked = False
output = ""
while not cracked:
    output = decrypt(packet, str(i))
    i += 1
    if output.startswith("GET") and output.endswith("HTTP/1.1"):
        print(output)
        cracked = True
```

b) Implementiere den Dijkstra-Algorithmus in Python und vergleiche deine Lösung.

```python
def dijkstra(graph, start_vertex):
    distances = dict()
    for v in graph:
        distances[v] = float("inf")
    distances[start_vertex] = 0
    unvisited =set(graph.keys())
    while unvisited:
        current_vertex = None
        min_distance = float("inf")
        for vertex in unvisited:
            if distances[vertex] < min_distance:    
                min_distance = distances[vertex]
                current_vertex = vertex
        unvisited.remove(current_vertex)
        
        for u in graph[current_vertex]:
            weight = distances[current_vertex] + graph[current_vertex][u]
            if distances[u] > weight:
                distances[u] = weight
                
        
    
    return distances
    
    
```

```python
graph = {
    'A': {'B': 4, 'C': 2},
    'B': {'A': 4, 'D': 3, 'E': 1, 'C': 1},
    'C': {'A': 2, 'B': 1, 'D': 5},
    'D': {'E': 2, 'C': 5, 'B':3},
    'E': {'D': 2, 'B': 1}
}

shortest_distances = dijkstra(graph, 'E')
print(shortest_distances)
```
