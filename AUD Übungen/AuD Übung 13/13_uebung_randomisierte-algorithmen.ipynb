## Aufgabe 2: Wurzelberechnung

Wir suchen nach effektiven Berechnungen für die Quadratwurzel einer Zahl. Da irrationale Zahlen unendlich viele Nachkommastellen haben können, versuchen wir die Ergebnisse bis zu einer bestimmten Genauigkeit zu ermitteln.

a) Nutze die Python-Bibliothek `random`, um mit Zufallssuche eine Lösung zu finden.

```python

import random

def g(a, x):
    return abs(a-x**2)

def sqrtRS(a, accuracy=0.0000001):
    x = random.uniform(0, a)
    gA = g(a, x)
    while gA > accuracy:
        x = random.uniform(0, a)
        gX = g(a, x)
        if gX < gA:
            gA = gX
    return x

x = sqrtRS(2)
print(x)
print(x ** 2)
```

b) Implementiere die Berechnung mittels des Bisektionsalgorithmus.

```python
def f(number, x):
    return number -x**2

def sqrtBS(n, a, b, accuracy=1e-12):
    fa = a
    fb = b
    while abs(fa - fb) > accuracy:
        t = (fa+fb)/2    
        if f(n, t) >0:
            fa = t
        else:
            fb = t
    return fa
x = sqrtBS(2, 0, 2)

print(x)
print(x**2)
```
