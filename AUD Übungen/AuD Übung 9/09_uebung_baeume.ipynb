```python
def decreaseKey(minHeap, index, newValue):
    if minHeap[index] >= newValue:
        minHeap[index] = newValue
        child = index
        heapCondition = False 
        while not heapCondition and child > 0:
            parent = int((child-1)/2)
            if minHeap[child] < minHeap[parent]:
                minHeap[parent], minHeap[child] = minHeap[child], minHeap[parent]
                child = parent
            else:
                heapCondition = True

def heapify(minHeap, index):
    lastInnerNode = int((len(minHeap)-2)/2)
    parent = index
    heapCondition = False
    while parent <= lastInnerNode and not heapCondition:
        child1 = parent*2+1
        child2 = parent*2+2
        print(parent,child1,child2)
        heapCondition = True
        newParent = parent
        if minHeap[child1] < minHeap[newParent]:
            newParent = child1
            heapCondition = False
        if minHeap[child2] < minHeap[newParent]:
            newParent = child2
            heapCondition = False
        minHeap[parent], minHeap[newParent] = minHeap[newParent], minHeap[parent]
        parent = newParent

def transformToHeap(minHeap):
    lastInnerNode = int((len(minHeap)-2)/2)
    for index in range(lastInnerNode,- 1, -1):
        heapify(minHeap, index)



def insert(minHeap, newValue):
    minHeap.append(newValue)
    transformToHeap(minHeap)

def delete(minHeap, index):
    del(minHeap[index])
    #oder minHeap.delete(index) --> ist eine funktion für Löschen in Sets
    transformToHeap(minHeap)

def getMin(minHeap):
    return minHeap[0]

def extractMin(minHeap):
    value = getMin(minHeap)
    minHeap.delete(minHeap[0])
    return value
```
